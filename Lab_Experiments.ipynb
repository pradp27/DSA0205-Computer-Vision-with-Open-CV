```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read image
img = cv2.imread("image.jpg")

# Check if image is loaded
if img is None:
    print("Error: Image not found!")
else:
    # Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Create kernel
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

    # Apply erosion
    eroded_img = cv2.erode(gray, kernel, iterations=1)

    # Display image
    cv2_imshow(eroded_img)

    # Save image
    cv2.imwrite("eroded_image.jpg", eroded_img)

    print("Eroded image saved successfully.")
```

```python
import cv2
from google.colab.patches import cv2_imshow

cap = cv2.VideoCapture("video.mp4")

count = 0

if not cap.isOpened():
    print("Error: Cannot open video.")
else:
    while count < 2:
        ret, frame = cap.read()

        if not ret:
            break

        slow_frame = cv2.resize(frame, None, fx=0.5, fy=0.5)
        fast_frame = cv2.resize(frame, None, fx=2.0, fy=2.0)

        print(f"Frame {count + 1} - Slow Motion")
        cv2_imshow(slow_frame)

        print(f"Frame {count + 1} - Fast Motion")
        cv2_imshow(fast_frame)

        input("Press Enter for next frame...")

        count += 1

cap.release()
print("Video processing completed.")
```

```python
import cv2
from google.colab.patches import cv2_imshow

# Upload video.mp4 to Colab
cap = cv2.VideoCapture("video.mp4")

slow_factor = 0.5
fast_factor = 2.0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    # Slow motion frame
    slow_frame = cv2.resize(
        frame,
        None,
        fx=slow_factor,
        fy=slow_factor
    )

    # Fast motion frame
    fast_frame = cv2.resize(
        frame,
        None,
        fx=fast_factor,
        fy=fast_factor
    )

    print("Slow Motion")
    cv2_imshow(slow_frame)

    print("Fast Motion")
    cv2_imshow(fast_frame)

    key = cv2.waitKey(100)

    if key == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read the image
img = cv2.imread('image.jpg')

# Get image dimensions
rows, cols = img.shape[:2]

# Define three points in the original image
pts1 = np.float32([
    [50, 50],
    [200, 50],
    [50, 200]
])

# Define the corresponding points in the transformed image
pts2 = np.float32([
    [10, 100],
    [200, 50],
    [100, 250]
])

# Compute the affine transformation matrix
M = cv2.getAffineTransform(pts1, pts2)

# Apply the transformation
out = cv2.warpAffine(img, M, (cols, rows))

# Display the original image
print("Original Image")
cv2_imshow(img)

# Display the transformed image
print("Affine Transformed Image")
cv2_imshow(out)

# Save the output image
cv2.imwrite("affine_transformed.jpg", out)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read the image
img = cv2.imread('image.jpg')

# Get image dimensions
h, w = img.shape[:2]

# Define the four corner points of the original image
src = np.float32([
    [0, 0],      # Top-left
    [w, 0],      # Top-right
    [0, h],      # Bottom-left
    [w, h]       # Bottom-right
])

# Define the destination points
dst = np.float32([
    [0, 0],         # Top-left
    [w, 0],         # Top-right
    [100, h],       # Bottom-left shifted right
    [w - 100, h]    # Bottom-right shifted left
])

# Compute the perspective transformation matrix
M = cv2.getPerspectiveTransform(src, dst)

# Apply the perspective transformation
out = cv2.warpPerspective(img, M, (w, h))

# Display the original image
print("Original Image")
cv2_imshow(img)

# Display the transformed image
print("Perspective Transformed Image")
cv2_imshow(out)

# Save the output image
cv2.imwrite("perspective_transformed.jpg", out)

print("Output image saved as perspective_transformed.jpg")
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read the video
cap = cv2.VideoCapture("video.mp4")

# Read the first frame to get video dimensions
ret, frame = cap.read()

if not ret:
    print("Error: Cannot read video.")
else:
    h, w = frame.shape[:2]

    # Source points (Corners of the frame)
    src = np.float32([
        [0, 0],
        [w, 0],
        [0, h],
        [w, h]
    ])

    # Destination points (Perspective effect)
    dst = np.float32([
        [100, 0],
        [w - 100, 0],
        [0, h],
        [w, h]
    ])

    # Perspective transformation matrix
    M = cv2.getPerspectiveTransform(src, dst)

    frame_count = 0
    max_frames = 5      # Display only first 5 frames

    while frame_count < max_frames:

        transformed = cv2.warpPerspective(frame, M, (w, h))

        print(f"Original Frame {frame_count + 1}")
        cv2_imshow(frame)

        print(f"Perspective Transformed Frame {frame_count + 1}")
        cv2_imshow(transformed)

        ret, frame = cap.read()

        if not ret:
            break

        frame_count += 1

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read the image
img = cv2.imread('image.jpg')

# Get image dimensions
h, w = img.shape[:2]

# Define source points
src = np.float32([
    [0, 0],      # Top-left
    [w, 0],      # Top-right
    [0, h],      # Bottom-left
    [w, h]       # Bottom-right
])

# Define destination points
dst = np.float32([
    [0, 0],          # Top-left
    [w, 0],          # Top-right
    [0, 0.7 * h],   # Bottom-left moved upward
    [w, 0.7 * h]    # Bottom-right moved upward
])

# Compute the homography matrix
H, status = cv2.findHomography(src, dst)

# Apply homography transformation
out = cv2.warpPerspective(img, H, (w, h))

# Display the original image
print("Original Image")
cv2_imshow(img)

# Display the transformed image
print("Homography Transformed Image")
cv2_imshow(out)

# Save the output image
cv2.imwrite("homography_output.jpg", out)

print("Output saved as homography_output.jpg")
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read the image
img = cv2.imread('image.jpg')

# Get image dimensions
h, w = img.shape[:2]

# Define source points
src_pts = np.float32([
    [0, 0],
    [w - 1, 0],
    [0, h - 1],
    [w - 1, h - 1]
])

# Define destination points
dst_pts = np.float32([
    [50, 50],
    [w - 100, 30],
    [100, h - 50],
    [w - 50, h - 100]
])

# Compute homography matrix using DLT
H, status = cv2.findHomography(
    src_pts,
    dst_pts,
    method=cv2.RANSAC
)

# Apply the transformation
transformed = cv2.warpPerspective(
    img,
    H,
    (w, h)
)

# Display original image
print("Original Image")
cv2_imshow(img)

# Display transformed image
print("DLT Transformed Image")
cv2_imshow(transformed)

# Save the output image
cv2.imwrite(
    "DLT_transformed_image.jpg",
    transformed
)

print("Image saved successfully.")
```
