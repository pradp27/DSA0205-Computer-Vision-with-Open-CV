```python
from google.colab.patches import cv2_imshow
import cv2

img = cv2.imread("image.jpg")

smoothed_img = cv2.blur(img, (5,5))

print("Original Image")
cv2_imshow(img)

print("Image after Smoothing (Average Filter 5x5)")
cv2_imshow(smoothed_img)
```

```python
from google.colab.patches import cv2_imshow
import cv2
import numpy as np

img = cv2.imread("image.jpg")

kernel = np.array([[0, -1, 0],
                   [-1, 5, -1],
                   [0, -1, 0]])

sharpened_img = cv2.filter2D(img, -1, kernel)

print("Original Image")
cv2_imshow(img)
print("Image after Sharpening")
cv2_imshow(sharpened_img)
```

```python
from google.colab.patches import cv2_imshow
import cv2

img = cv2.imread("image.jpg")

mean_img = cv2.blur(img, (5,5))

print("Original Image")
cv2_imshow(img)

print("Image after Mean Filtering")
cv2_imshow(mean_img)
```

```python
from google.colab.patches import cv2_imshow
import cv2

img = cv2.imread("image.jpg")

median_img = cv2.medianBlur(img, 5)

print("Original Image")
cv2_imshow(img)

print("Image after Median Filtering")
cv2_imshow(median_img)
```

```python
from google.colab.patches import cv2_imshow
import cv2

img = cv2.imread("image.jpg")

gaussian_img = cv2.GaussianBlur(img, (5,5), 0)

print("Original Image")
cv2_imshow(img)

print("Image after Gaussian Filtering")
cv2_imshow(gaussian_img)
```
